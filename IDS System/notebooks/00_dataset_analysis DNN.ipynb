```python
import pandas as pd

chunk_size = 200000
chunks = []

for chunk in pd.read_csv("../data/raw/DNN-EdgeIIoT-dataset.csv", chunksize=chunk_size):
    chunks.append(chunk)

df_ml = pd.concat(chunks)

print(df_ml.shape)
```

```python
print("Columns:")
print(df_ml.columns.tolist())
```

```python
print(df_ml["Attack_type"].value_counts())
print("\nPercentages:")
print(df_ml["Attack_type"].value_counts(normalize=True))
```

```python
if "Attack_label" in df_ml.columns:
    print(df_ml["Attack_label"].value_counts())
```

```python
print("Total feature columns:", len(df_ml.columns) - 1)
```

```python
import numpy as np

df_numeric = df_ml.select_dtypes(include=[np.number])

correlations = df_numeric.corr()["Attack_label"] if "Attack_label" in df_numeric.columns else None

print(correlations.sort_values(ascending=False).head(10))
```

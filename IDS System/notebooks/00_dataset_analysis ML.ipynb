```python
import pandas as pd

df_ml = pd.read_csv("../data/raw/ML-EdgeIIoT-dataset.csv", low_memory=False)

print("Shape:", df_ml.shape)
print("Memory usage (MB):", df_ml.memory_usage(deep=True).sum() / (1024**2))
df_ml.head()
```

```python
print("Columns:")
print(df_ml.columns.tolist())
```

```python
print(df_ml["Attack_type"].value_counts())
print("\nPercentages:")
print(df_ml["Attack_type"].value_counts(normalize=True))
```
